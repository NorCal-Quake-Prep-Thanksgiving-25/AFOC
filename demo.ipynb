# Autonomous Fiscal Orchestration Core Demo

Demonstrates the six-model pipeline: governance, forecasting, optimisation, and reporting.

```python
from afoc import ComposableIntelligenceCore
from afoc.datatypes import StrategicAnalysis, StrategicPlan, StrategicRoadmap

core = ComposableIntelligenceCore()
analysis = StrategicAnalysis(
    feasibility_score=0.92,
    alignment_vector={'architecture': 0.4, 'implementation': 0.35, 'optimization': 0.25},
    narrative='Launch enterprise-grade automation platform',
)
governance = core.establish_fiscal_governance(analysis)
roadmap = StrategicRoadmap(
    milestones=['baseline', 'scale'],
    fiscal_targets={'m0': 1200.0, 'm1': 1500.0, 'm2': 1900.0, 'm3': 2300.0},
    scenario_assumptions={'seasonality': 1.0},
    growth_projection=1.08,
)
forecast = core.predictive_fiscal_forecasting(roadmap)
optimisation = core.execute_fiscal_optimization_cycle({'architecture': 0.8, 'implementation': 0.6, 'optimization': 0.4})
governance.summary(), forecast.mean, optimisation.selected_action
```

```python
from pathlib import Path
from afoc.datatypes import StrategicPlan
from afoc.tools import export_pdf_summary

plan = StrategicPlan(
    architecture_budget=120000.0,
    implementation_budget=240000.0,
    optimization_budget=80000.0,
    architectural_tasks=['Refactor core'],
    implementation_tasks=['Deploy integrations'],
    optimization_tasks=['Tune ML lifecycle'],
    technical_constraints=['Budget'],
    fiscal_constraints={'roi_floor': 1.5},
    fiscal_guidelines={'spend_vs_budget': 1.0},
    performance_slas={'latency': 0.2},
    roi_targets={'year1': 1.4},
    coding_standards=['PEP8'],
)
dashboard = core.monitor_fiscal_operations(plan)
output = export_pdf_summary(governance=governance, dashboard=dashboard, output_path=Path('afoc_report.pdf'))
output
```
